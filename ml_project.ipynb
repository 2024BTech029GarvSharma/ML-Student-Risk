```python
# Define model configurations for all strategies
models = {
    "Linear Regression": {
        "model": LinearRegression(),
        "params": {}
    },
    "DecisionTree": {
        "model": DecisionTreeRegressor(),
        "params": {"max_depth": [3, 5, 10], "min_samples_split": [2, 5]}
    },
    "RandomForest": {
        "model": RandomForestRegressor(),
        "params": {"n_estimators": [50, 100], "max_depth": [5, 10]}
    },
    "XGBoost": {
        "model": XGBRegressor(random_state=42),
        "params": {"n_estimators": [50, 100, 150], "learning_rate": [0.01, 0.1], "max_depth": [3, 5, 7]}
    }
}

# Also define common metrics for convenience
from sklearn.metrics import mean_absolute_error
```

```python
# EXPORT PREDICTIONS - BEST MODEL FROM ENSEMBLE STRATEGY
print("\n" + "="*100)
print("FINAL SELECTION & PREDICTION EXPORT")
print("="*100)

# Ensure all strategy dataframes exist
if 'ensemble_results_df' not in locals() or len(ensemble_results_df) == 0:
    print("\n⚠️  Running Strategy 4 (Ensemble) first...")

    # Quick ensemble setup
    df_all = df.copy()
    categorical_features = ["gender","part_time_job","diet_quality","parental_education_level","internet_quality","extracurricular_participation"]

    for col in categorical_features:
        le = LabelEncoder()
        df_all[col] = le.fit_transform(df_all[col])

    all_numeric_features = df_all.select_dtypes(include=[np.number]).columns.tolist()
    all_numeric_features.remove('exam_score')
    if 'student_id' in all_numeric_features:
        all_numeric_features.remove('student_id')

    X_all = df_all[all_numeric_features]
    y_all = df_all['exam_score']

    # Get ensemble features (combination of all methods)
    from sklearn.feature_selection import mutual_info_regression

    mi_scores = mutual_info_regression(X_all, y_all, random_state=42)
    mi_importance = pd.DataFrame({'Feature': all_numeric_features, 'MI_Score': mi_scores}).sort_values('MI_Score', ascending=False)
    mi_top_features = mi_importance.head(5)['Feature'].tolist()

    rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
    rf.fit(X_all, y_all)
    feature_importance = pd.DataFrame({'Feature': all_numeric_features, 'Importance': rf.feature_importances_}).sort_values('Importance', ascending=False)
    rf_top_features = feature_importance.head(5)['Feature'].tolist()

    correlation_with_target = df_all.corr(numeric_only=True)['exam_score'].sort_values(ascending=False)
    corr_top_features = correlation_with_target[1:6].index.tolist()

    # Ensemble logic
    feature_scores = {}
    for feature in mi_top_features + rf_top_features + corr_top_features:
        feature_scores[feature] = feature_scores.get(feature, 0) + 1

    ensemble_features = sorted(feature_scores.items(), key=lambda x: x[1], reverse=True)
    ensemble_top_features = [f[0] for f in ensemble_features[:6]]

    # Prepare ensemble data
    X_ensemble = df_all[ensemble_top_features]
    y_ensemble = df_all['exam_score']
    X_train_ens, X_test_ens, y_train_ens, y_test_ens = train_test_split(X_ensemble, y_ensemble, test_size=0.2, random_state=42, shuffle=True)

    # Train best ensemble model
    ensemble_results = []
    for name, config in models.items():
        grid = GridSearchCV(config["model"], config['params'], cv=5, scoring="neg_mean_squared_error")
        grid.fit(X_train_ens, y_train_ens)

        y_pred_ens = grid.predict(X_test_ens)
        rmse_ens = np.sqrt(mean_squared_error(y_test_ens, y_pred_ens))
        mae_ens = mean_absolute_error(y_test_ens, y_pred_ens)
        r2_ens = r2_score(y_test_ens, y_pred_ens)

        ensemble_results.append({
            'Strategy': 'Ensemble Mixed',
            'Model': name,
            'Features': ', '.join(ensemble_top_features),
            'RMSE': rmse_ens,
            'MAE': mae_ens,
            'R² Score': r2_ens,
            'Predictions': y_pred_ens
        })

    ensemble_results_df = pd.DataFrame(ensemble_results)
    print("✓ Ensemble strategy created automatically")

# Select best model from ensemble strategy (most robust)
best_ensemble = ensemble_results_df.loc[ensemble_results_df['R² Score'].idxmax()]

print(f"\n✓ SELECTED FINAL MODEL:")
print(f"  Strategy: {best_ensemble['Strategy']}")
print(f"  Model: {best_ensemble['Model']}")
print(f"  Features: {best_ensemble['Features']}")
print(f"  RMSE: {best_ensemble['RMSE']:.4f}")
print(f"  MAE: {best_ensemble['MAE']:.4f}")
print(f"  R² Score: {best_ensemble['R² Score']:.4f}")

# Get predictions from best model
best_predictions = best_ensemble['Predictions']

print(f"\n📊 Prediction Statistics:")
print(f"  Min Prediction: {best_predictions.min():.2f}")
print(f"  Max Prediction: {best_predictions.max():.2f}")
print(f"  Mean Prediction: {best_predictions.mean():.2f}")
print(f"  Std Prediction: {best_predictions.std():.2f}")

# Create comparison dataframe
prediction_comparison = pd.DataFrame({
    'Actual': y_test_ens.values,
    'Predicted': best_predictions,
    'Error': y_test_ens.values - best_predictions,
    'Abs_Error': np.abs(y_test_ens.values - best_predictions)
})

print(f"\n📋 Sample Predictions (First 10):\n")
print(prediction_comparison.head(10).to_string(index=False))

print("\n✓ Prediction Export Complete!")
print("="*100)
```

```python
# STRATEGY 4: ENSEMBLE/MIXED APPROACH - INTELLIGENT COMBINATION
print("\n" + "="*100)
print("STRATEGY 4: ENSEMBLE/MIXED APPROACH - SMART FEATURE COMBINATION")
print("="*100)

# Ensure df_mi exists (prepare encoded data)
if 'df_mi' not in locals():
    df_mi = df.copy()
    categorical_features = ["gender","part_time_job","diet_quality","parental_education_level","internet_quality","extracurricular_participation"]

    for col in categorical_features:
        le = LabelEncoder()
        df_mi[col] = le.fit_transform(df_mi[col])

# Ensure X_all and y_all exist for feature selection
if 'X_all' not in locals():
    all_numeric_features = df_mi.select_dtypes(include=[np.number]).columns.tolist()
    all_numeric_features.remove('exam_score')
    if 'student_id' in all_numeric_features:
        all_numeric_features.remove('student_id')

    X_all = df_mi[all_numeric_features]
    y_all = df_mi['exam_score']

# Get features from each method if not already calculated
if 'mi_top_features' not in locals():
    from sklearn.feature_selection import mutual_info_regression
    mi_scores = mutual_info_regression(X_all, y_all, random_state=42)
    mi_importance = pd.DataFrame({
        'Feature': all_numeric_features,
        'MI_Score': mi_scores
    }).sort_values('MI_Score', ascending=False)
    mi_top_features = mi_importance.head(5)['Feature'].tolist()

if 'rf_top_features' not in locals():
    rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
    rf.fit(X_all, y_all)
    feature_importance = pd.DataFrame({
        'Feature': all_numeric_features,
        'Importance': rf.feature_importances_
    }).sort_values('Importance', ascending=False)
    rf_top_features = feature_importance.head(5)['Feature'].tolist()

if 'corr_top_features' not in locals():
    correlation_with_target = df_mi.corr(numeric_only=True)['exam_score'].sort_values(ascending=False)
    corr_top_features = correlation_with_target[1:6].index.tolist()

print("\nFeature Selection Methods:")
print(f"  MI Top 5:           {mi_top_features}")
print(f"  RF Importance Top 5: {rf_top_features}")
print(f"  Correlation Top 5:   {corr_top_features}")

# Count feature appearances across methods
feature_scores = {}
for feature in mi_top_features:
    feature_scores[feature] = feature_scores.get(feature, 0) + 1
for feature in rf_top_features:
    feature_scores[feature] = feature_scores.get(feature, 0) + 1
for feature in corr_top_features:
    feature_scores[feature] = feature_scores.get(feature, 0) + 1

# Sort by score (how many methods selected it) then by position
ensemble_features = sorted(feature_scores.items(), key=lambda x: x[1], reverse=True)
ensemble_top_features = [f[0] for f in ensemble_features[:6]]  # Top 6 features

print(f"\n✓ Ensemble Feature Selection Logic:")
print(f"  Features appearing in multiple methods get priority")
print(f"  Selected Ensemble Features: {ensemble_top_features}")
print(f"  Feature Coverage Score: {[dict(ensemble_features)[f] for f in ensemble_top_features]}")

# Prepare data with ensemble features
X_ensemble = df_mi[ensemble_top_features]
y_ensemble = df_mi['exam_score']

# Train-test split with shuffle
X_train_ens, X_test_ens, y_train_ens, y_test_ens = train_test_split(X_ensemble, y_ensemble, test_size=0.2, random_state=42, shuffle=True)

print(f"\nTrain size: {len(X_train_ens)}, Test size: {len(X_test_ens)}")

# Train all 4 models on ensemble features
ensemble_results = []

for name, config in models.items():
    grid = GridSearchCV(config["model"], config['params'], cv=5, scoring="neg_mean_squared_error")
    grid.fit(X_train_ens, y_train_ens)

    y_pred_ens = grid.predict(X_test_ens)

    rmse_ens = np.sqrt(mean_squared_error(y_test_ens, y_pred_ens))
    mae_ens = mean_absolute_error(y_test_ens, y_pred_ens)
    r2_ens = r2_score(y_test_ens, y_pred_ens)

    ensemble_results.append({
        'Strategy': 'Ensemble Mixed',
        'Model': name,
        'Features': ', '.join(ensemble_top_features),
        'RMSE': rmse_ens,
        'MAE': mae_ens,
        'R² Score': r2_ens,
        'Predictions': y_pred_ens
    })

    print(f"{name:20} | RMSE: {rmse_ens:.4f} | MAE: {mae_ens:.4f} | R²: {r2_ens:.4f}")

ensemble_results_df = pd.DataFrame(ensemble_results)
print("\n✓ Ensemble Strategy Complete")
```

```python
# STRATEGY 2: RANDOM FOREST FEATURE IMPORTANCE
print("\n" + "="*100)
print("STRATEGY 2: RANDOM FOREST FEATURE IMPORTANCE - TOP 5 FEATURES")
print("="*100)

# Ensure RF importance data exists
if 'feature_importance' not in locals():
    rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
    rf.fit(X_all, y_all)

    feature_importance = pd.DataFrame({
        'Feature': all_numeric_features,
        'Importance': rf.feature_importances_
    }).sort_values('Importance', ascending=False)

# Get top 5 RF importance features
rf_top_features = feature_importance.head(5)['Feature'].tolist()
print(f"\nSelected Features (RF Importance): {rf_top_features}")

# Prepare data with RF features
X_rf = df_mi[rf_top_features]
y_rf = df_mi['exam_score']

# Train-test split with shuffle
X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_test_split(X_rf, y_rf, test_size=0.2, random_state=42, shuffle=True)

print(f"Train size: {len(X_train_rf)}, Test size: {len(X_test_rf)}")

# Train all 4 models on RF importance features
rf_results = []

for name, config in models.items():
    grid = GridSearchCV(config["model"], config['params'], cv=5, scoring="neg_mean_squared_error")
    grid.fit(X_train_rf, y_train_rf)

    y_pred_rf = grid.predict(X_test_rf)

    rmse_rf = np.sqrt(mean_squared_error(y_test_rf, y_pred_rf))
    mae_rf = mean_absolute_error(y_test_rf, y_pred_rf)
    r2_rf = r2_score(y_test_rf, y_pred_rf)

    rf_results.append({
        'Strategy': 'RF Importance',
        'Model': name,
        'Features': ', '.join(rf_top_features),
        'RMSE': rmse_rf,
        'MAE': mae_rf,
        'R² Score': r2_rf,
        'Predictions': y_pred_rf
    })

    print(f"{name:20} | RMSE: {rmse_rf:.4f} | MAE: {mae_rf:.4f} | R²: {r2_rf:.4f}")

rf_results_df = pd.DataFrame(rf_results)
print("\n✓ Random Forest Importance Strategy Complete")
```
